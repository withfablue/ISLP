```python
college.describe()
```

```python
pd.plotting.scatter_matrix(college[['Top10perc', 'Apps', 'Enroll']])
```

```python
from matplotlib.pyplot import subplots
fig, ax = subplots( figsize=(5,5))
college.boxplot('Outstate', by='Private', ax=ax)
```

```python
college['Elite'] = pd.cut(college['Top10perc']/100,
                          [0, 0.5, 1],
                          labels=['No', 'Yes'])

college['Elite'].value_counts()
```

```python
fig, ax = subplots(figsize=(5,5))
college.boxplot('Outstate', by='Elite', ax=ax)
```

```python
fig, axes = subplots(2, 2, figsize=(16, 16))

college.hist('Enroll', color='blue', bins=7, ax=axes[0,0])
college.hist('PhD', color='green', bins=5, ax=axes[0,1])
college.hist('Outstate', color='gray', bins=10, ax=axes[1,0])
college.hist('Apps', color='yellow', bins=5, ax=axes[1,1])
```
